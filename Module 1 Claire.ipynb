* **Standard of care treatments (& reimbursement)**
    * Anti-amyloid immunotherapy is used in early stages of Alzheimer's disease to target cognitive decline from amyloid plaques. Monoclonal antibodies like lecanemab and donanemab are examples that have been shown to slow Alzheimer's symptoms (Wu & Fuh, 2025).
    * Symptomatic relief for mild to severe Alzheimer's is typically provided by cholinesterase inhibitors and memantine which prevent the neurotransmitter acetylcholine from breaking down (Parsons et. al, 2013). Increased acetylcholine helps relieve dementia effects because acetylcholine allows the patient to not forget short-term memories as often and helping the brain to better organize long-term memory (Mental Health America, n.d.).
* **Disease progression & prognosis**
    * According to Mayo Clinic (2025), the following is the progression and prognosis for Alzheimer's disease:
    * Alzheimer's disease in the early stage is hard to diagnose. Biomarkers like amyloid plaques and tau tangles appear in brain scans, but symptoms are not yet noticed. This stage can last many years, but technology is evolving to help diagnose Alzheimer's earlier.
    * Alzheimer's can cause mild cognitive impairment (MCI) which doesn't directly affect the patient's work or relationships, but affects their ability to recall things that they normally would be able to easily. Having MCI doesn't necessarily indicate that a person has Alzheimer's, but it can be a sign that can assist in an early diagnosis through checking for biomarkers.
    * The diagnosis of Alzheimer's disease typically occurs during the mild dementia stage. This is when people close to the patient can begin to pick up on their dementia. People in this stage experience short-term memory loss, personality changes, trouble thinking clearly, and getting lost or losing items.
    * Moderate dementia during Alzheimer's is when people forget even more and become more confused. They begin to need help with their everyday life, have poor judgment, and experience greater personality changes.
    * Severe dementia occurs when the patient begins to lose their physical abilities, such as the ability to walk. The patient now needs assistance with their everyday life and may even lose the ability to communicate.
    * People with Alzheimer's disease typically live for around 3 to 11 years after diagnosis.
* **Continuum of care providers**
    * The following information comes from the National Council of Certified Dementia Practicioners (n.d.):
    * In early stages of Alzheimer's, no help for medical professionals is required, as most of the time, Alzheimer's is not diagnosed yet. Alzheimers is typically diagnosed in the mild dementia stage, which can last from 2 to 4 years.
    * After the mild dementia stage, when the patient is experiencing moderate dementia, they will begin to need help with daily activities.
    * During the moderately severe dementia stage, the patient may begin to also require assistance with things like "dressing, bathing, and toileting."
    * In the final stage of dementia, the patient can no longer walk or sit without help. This is the point where the individual fully depends on care providers.
* **Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)**
    * The root causes of Alzheimer's are not completely known, researchers recognize two main proteins as playing an influential role in the disease. Amyloid-beta fragments in the brain form into plaques and Tau proteins assemble into tangles, which can damage neurons in the memory part of the brain or inhibit their communication. This damage can spread throughout the brain and cause the individual to lose their memory and their ability to function. The damage spreading through the brain can also cause brain shrinkage, meaning a lower brain weight (Mayo Clinic, 2026).
* **Clinical Trials/next-gen therapies**
    * Currently, there are many ongoing clinical trials for the prevention/mediation of Alzheimer's disease. Some of these clinical trials are focused on disease-modifying therapies, which aim to stop the causes by targeting amyloid plaques and "change the course of AD." Other clinical trials are symptomatic, focused on fixing cognitive impairment or other symptoms of dementia. There are also therapies in trial that focus specifically on reducing tau or amyloid beta to slow the formation of tangles or plaques in the brain (Huang et. al, 2023).

## Data Analysis: 
First we created a Patient object that stored each donors information including ID, sex, age at death, cognitive status, pTAU level, amyloid-beta 42 level, and brain weight.

```python
import csv

class Patient:
    all_patients = []
    # the following lines assign attributes to each patient object 
    def __init__(self, id: str, sex: str, age_at_death: float, cog_status: str, ptau_level: float, abeta42_level: float, brain_weight: int | None):
        self.id = id
        self.sex = sex
        self.age_at_death = age_at_death
        self.cog_status =  cog_status
        self.ptau_level = ptau_level
        self.abeta42_level = abeta42_level
        self.brain_weight = brain_weight
        Patient.all_patients.append(self) # add the patient to the list of patients
    
    # the following lines set how the patient will be displayed as text when printed
    def __repr__(self): 
        return f"Patient {self.id}: (Sex: {self.sex} | Age at Death: {self.age_at_death} | Cognitive Status: {self.cog_status} | Aβ42: {self.abeta42_level} pg/ug | pTAU: {self.ptau_level} pg/ug | Brain Weight: {self.brain_weight} g)"

    # retrieves patient brain weight
    def get_brain_weight(self):
        return self.brain_weight

    # the following lines read the data from the csv file and create patient objects
    @classmethod
    def instantiate_from_csv(cls, filename: str):
        with open(filename, encoding = "utf8") as f:
            reader = csv.DictReader(f)
            rows_of_patients = list(reader)
            # iterates through each row of patient data
        for row in rows_of_patients:
            Patient(
                id = row['Donor ID'],
                sex = row['Sex'],
                age_at_death = float(row['Age at Death']),
                cog_status = row['Cognitive Status'],
                ptau_level = float(row['pTAU pg/ug']),
                abeta42_level = float(row['ABeta42 pg/ug']),
                brain_weight = int(row['Fresh Brain Weight']) if row['Fresh Brain Weight'] != 'Unavailable' else None
            )   
    # retrieve patient data with their id   
    @classmethod
    def get_patient(cls, id):
        for patient in Patient.all_patients:
            if id == patient.id: 
                return patient
    # the following lines help to filter patients based on given attributes            
    @classmethod
    def filter(cls, list, id:str = "any",sex:str ="any", age_at_death:int ="any", cog_status:str ="any", ptau_level:int ="any", abeta42_level:int ="any", brain_weight:int ="any"):
        all_patients = list
        remove_list = []
        # grouping attribute names into tuples
        attr_list = (id, sex, age_at_death, cog_status, ptau_level, abeta42_level)
        attr_name = ("id", "sex", "age_at_death", "cog_status", "ptau_level", "abeta42_level")
    # loop through attributes to remove unwanted objects
        for attr in range(len(attr_list)):
            if attr_list[attr] != 'any':
                for patient in all_patients:
                    if getattr(patient, attr_name[attr]) != attr_list[attr]:
                        remove_list.append(patient)
            all_patients = [patient for patient in all_patients if patient not in remove_list]
            remove_list.clear()
        return all_patients
```

We then used a T-test with a bar graph comparing donor brain weight and cognitive status. This yielded a p value of around 0.02, less than 0.05 and therefore statistically significant. This makes sense because Alzheimer's can cause damage to neurons that overall leads to brain shrinkage.

```python
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics
from patient_Claire import * 
from sklearn.linear_model import LinearRegression

########## bar graph code: 
# Analyzing brain weight between dementia and no dementia patients

# start with empty lists for brain weight with dementia and no dementia respectively
brain_weight_dementia = []
brain_weight_no_dementia = []

for patient in Patient.filter(Patient.all_patients, cog_status = "Dementia"):
    if patient.brain_weight is not None:
        brain_weight_dementia.append(patient.brain_weight)
for patient in Patient.filter(Patient.all_patients, cog_status = "No dementia"):
    if patient.brain_weight is not None:
        brain_weight_no_dementia.append(patient.brain_weight)

# calculating the mean of each data set for dementia/no dementia
x_dementia_bar = statistics.mean(brain_weight_dementia)
x_no_dementia_bar = statistics.mean(brain_weight_no_dementia)

# calculating the standard deviation for dementia/no dementia
brain_weight_dementia_stdev = statistics.stdev(brain_weight_dementia)
brain_weight_no_dementia_stdev = statistics.stdev(brain_weight_no_dementia)

# print calculations
print(f'x_dementia_bar = {x_dementia_bar}, brain_weight_dementia_stdev {brain_weight_dementia_stdev}')
print(f'x_no_dementia_bar = {x_no_dementia_bar}, brain_weight_no_dementia_stdev {brain_weight_no_dementia_stdev}')

# setting up axis labels, heights, error bar length
patient_cog_status_cols = ['Dementia', 'No Dementia']
mean_brain_weight = [x_dementia_bar, x_no_dementia_bar]
stdev_brain_weight = [brain_weight_dementia_stdev, brain_weight_no_dementia_stdev]

yerr = [np.zeros(len(mean_brain_weight)), stdev_brain_weight] # sets the error bottom as the top of the bar


t_stat, p_val = stats.ttest_ind(brain_weight_dementia, brain_weight_no_dementia)
print(f't_stat =  {t_stat}, p_val = {p_val}') # finding the p value and t statistic

# plot graph with colors and error bars
plt.bar(patient_cog_status_cols, mean_brain_weight, yerr = yerr, capsize=10, color=["pink", "blue"])
plt.title("Average Brain Weight by Cognitive Status") # set title

#labeling axises
plt.xlabel("Cognitive Status")
plt.ylabel("Average Brain Weight (g)")

# set y-axis range to 1500 so bars and error caps have plenty of room
plt.ylim(0, 1500)

# Position text of p value and t stat 
plt.text(
        0.5, 0.92,
        f"t = {t_stat:.2f}\np = {p_val:.3e}",
        ha = 'center',
        va = 'top',
        transform = plt.gca().transAxes
        )
plt.show()
```

As we can see above, the donors without dementia had higher brain weights on average and with statistical significance.

Next, we made a scatterplot comparing brain weight with amyloid beta 42 levels. The R value ended up being far below 1, so there was not that strong of a correlation between the two.

```python
########## scatter plot code: 
# Analyzing correlation between amyloid beta levels and brain weight
# creating empty lists for patients brain weight and amyloid beta levels
patient_abeta42 = []
patient_brain_weight = []

# populating the empty lists above
for patient in Patient.all_patients:
    if patient.abeta42_level is not None and patient.brain_weight is not None:
        patient_abeta42.append(patient.abeta42_level)
        patient_brain_weight.append(patient.brain_weight)

# defining x and y axises
X = patient_abeta42 # independent variable (amyloid beta level)
y = patient_brain_weight # dependent variable (brain weight)

array_x = np.asarray(patient_abeta42, dtype = float).reshape(-1, 1) # shape: (n_samples, 1)
array_y = np.asarray(patient_brain_weight, dtype = float).ravel() # shape: (n_samples,)

model = LinearRegression()
model.fit(array_x, array_y)

# Create predicted values for the regression line
predicted_y = model.predict(array_x)

# Sort x-values so the line displays correctly
sort_indices = np.argsort(array_x[:, 0])

# adding labels and titles
plt.scatter(array_x.ravel(), array_y, color='blue')
plt.plot(
    array_x[sort_indices].ravel(),
    predicted_y[sort_indices],
    color="red",
    linewidth=2,
    label="Linear regression"
) # adding the regression line

slope = model.coef_[0]
intercept = model.intercept_
r_squared = model.score(array_x, array_y)
plt.text(
    0.55, 0.95,
    f"y = {slope:.3f}x + {intercept:.3f}\nR² = {r_squared:.3f}",
    transform=plt.gca().transAxes,
    verticalalignment="top",
    fontsize=11
) # adding the R^2 value and the line equation

plt.xlabel('Aβ42 Level (pg/ug)')
plt.ylabel('Brain Weight (g)')
plt.title('Scatter Plot of Aβ42 Level vs Brain Weight')
plt.show()
```

The line points slightly downward, but there is a very weak relationship, since the R^2 value is 0.002. There is also one donor that has a much higher amyloid beta 42 level than anyone else, which may affect the data and its variance.
